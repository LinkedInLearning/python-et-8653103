```python
from urllib.request import urlopen, Request
import json

data = json.dumps({"key": "value"}).encode("utf-8")

req = Request(
    "https://httpbin.org/post",
    data=data,
    headers={"Content-Type": "application/json"}
)

with urlopen(req) as response:
    print(response.read().decode())
```
