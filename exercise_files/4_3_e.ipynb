```python
numbers = [3, 1, 4, 1, 5, 9]

def get_numbers():
    for n in numbers:
        yield n
        print(f'Yielded {n}')
```

```python
numbers_generator = get_numbers()
```

```python
type(numbers_generator)
```

```python
next(numbers_generator)
```

```python
def get_primes():
    primes = [2]
    for n in range(3, 1000000):
        for factor in primes:
            if factor > (n ** 0.5):
                primes.append(n)
                yield n
                break
            
            if n % factor == 0:
                break
```

```python
def infinite_range(start):
    n = start
    while True:
        yield n
        n += 1
```

```python
def get_primes():
    primes = [2]
    for n in infinite_range(3):
        for factor in primes:
            if factor > (n ** 0.5):
                primes.append(n)
                yield n
                break
            
            if n % factor == 0:
                break
```

```python
primes_generator = get_primes()
```

```python
next(primes_generator)
```
