```python
import csv

with open("airports.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    airport_types = {
        row["iata_code"]: row["type"]
        for row in reader
        if row["iata_code"]  # filter out rows with no IATA code
    }
```

```python
import pprint

def write_dict_as_module(d, filename, var_name="data"):
    with open(filename, "w", encoding="utf-8") as f:
        f.write(f"{var_name} = ")
        f.write(pprint.pformat(d))
        f.write("\n")
```

```python
write_dict_as_module(airport_types, 'data_3_2_airport_types')
```

## Keys, Values, and Items

```python
codes = ["BOS", "BLH", "SIN", "LHR", "PYT", "HON"]
```

```python
small = {}
for code in codes:
    small[code] = airports[code]
```

```python
small.keys()
```

```python
for item in small.items():
    print(f"Airport code {item[0]} is for {item[1]}")
```
