```python
from datetime import datetime

wait_until = datetime.now().second + 2

while datetime.now().second < wait_until:
	print("Waiting")
print(f"We are at {wait_until} seconds")
```

```python
import time

wait_until = datetime.now().second + 2

while datetime.now().second < wait_until:
    print("Waiting")
    time.sleep(0.5)

print(f"We are at {wait_until} seconds")
```

```python
wait_until = datetime.now().second + 2

while True:
    if datetime.now().second == wait_until:
        break

print(f"We are at {wait_until} seconds")
```

```python
wait_until = datetime.now().second + 2

while True:
    while datetime.now().second == wait_until:
        print(f"We are at {wait_until} seconds")
        time.sleep(0.5)
        break

print(f"Done!")
```
