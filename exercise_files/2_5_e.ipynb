```python
import os

def open_file(path):
    if os.path.exists(path):
        return open(path)
    return None

file = open_file('data_2_2.py')
if file:
    print('Do something with the file')
```

```python
file = open_file('does_not_exist.py')
if file:
    print('Do something with the file')
```
