```python
class Date:
  def __init__(self, month, day, year):
    self.month = month
    self.day = day
    self.year = year


  def get_formatted_string(self):
    return f"{self.month} {self.day} {self.year}"
```

```python
birthday = Date(2, 20, 1991)
```

```python
birthday.get_formatted_string()
```

```python
class Date:
  def __init__(self, month, day, year):
    self.month = month
    self.day = day
    self.year = year


  def get_formatted_string(self):
    return f"{calendar.month_name[self.month]} {self.day} {self.year}"
```

```python
birthday = Date(2, 20, 1991)
birthday.get_formatted_string()
```
