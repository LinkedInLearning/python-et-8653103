```python
from IPython.display import HTML, Image, display
```

```python
def thread(*args):
  print(type(args))
  print(args)
  num_posts = len(args)
  for i, post in enumerate(args):
    print(f"{post} ({i}/{num_posts} 🧵)")

thread("here's a story I'm going to tell", "About Python", "and use *args to do it!")
```

```python
template = "__ will help me __ my __ for __."

words = ["Python", "analyze", "data", "insights"]
```

```python
def word_fill_in(template, *words):
  template = template.replace("__", "{}")

  return template.format(*words)

word_fill_in(template, "Python", "analyze", "data", "insights")


```

```python
def print_keyword_arguments(**kwargs):
  print(type(kwargs))
  print(kwargs)

print_keyword_arguments(color="blue", size="small")
```

```python
def display_item(SKU, name, size="One Size", image="data_5_4.png", **kwargs):
  display(HTML(f"<h1>{name}</h1><h3>{SKU}</h3>"))
  product_info = [f"Size: {size}"] + [f"{key}: {val}" for key, val in kwargs.items()]
  display(HTML(f"<ul><li>{"</li><li>".join(product_info)}</li></ul>"))
  display(Image(filename="data_5_4.png"))
```

```python
display_item(12345, "Blue Shirt", color="blue", material="mithril", size="small")
```
