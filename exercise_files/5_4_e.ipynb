```python
def thread(*args):
    print(type(args))
    print(args)
    num_threads = len(args)

    for i, post in enumerate(args):
        print(f'{post} ({i+1}/{num_threads} 🧵)')


thread("Here is a long story I'm going to tell", "about Python", "I can use *args to do it!")
```

```python
def print_keyword_args(**kwargs):
    print(type(kwargs))
    print(kwargs)
```

```python
from IPython.display import HTML, Image, display


def display_item(SKU, name, *args, size="One Size", image="data_5_4.png", **kwargs):
    display(HTML(f"<h1>{name}</h1><h3>{SKU}</h3>"))
    product_info = [f'Size: {size}'] + [f'{key}: {val}' for key, val in kwargs.items()]
    display(HTML(f'<ul><li>{'</li><li>'.join(product_info)}</li></ul>'))
    display(Image(filename=image))

```

```python
display_item(12345, "Blue Shirt", color="blue", material="mithril", size="small")
```
