```python
def sum_of_squares(n1: int, n2: int) -> int:
	"""
	Return the sum of the squares of two integers
	"""
	if type(n1) != int or type(n2) != int:
		print("This function only works with integers!")
		return 

	return n1**2 + n2**2
```

```python
sum_of_squares("hello", "world")
```
