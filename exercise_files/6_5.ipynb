```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 500)
y = np.linspace(-5, 5, 500)
X, Y = np.meshgrid(x, y)
Z = (np.sin(np.sqrt(X**2 + Y**2)) + 
     np.sin(np.sqrt((X-2)**2 + (Y-2)**2)) +
     np.sin(np.sqrt((X+2)**2 + (Y-1)**2)))
plt.figure(figsize=(8, 8))
plt.contourf(X, Y, Z, levels=100, cmap='plasma')
plt.colorbar()
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from nltk import download as nltk_download

nltk_download('gutenberg')

from nltk.corpus import gutenberg


text = gutenberg.raw('melville-moby_dick.txt')
wc = WordCloud(width=800, height=400, background_color='black', colormap='plasma').generate(text)
plt.imshow(wc)
plt.axis('off')
plt.show()
```
