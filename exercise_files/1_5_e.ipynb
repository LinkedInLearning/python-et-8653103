```python
from datetime import datetime
```

```python
datetime.fromisoformat("2025-07-31T00:00:00")
```

```python
def format_date(date):
	return date.split("T")

# On July 7, 2025 it was 90.0 degrees Fahrenheit
def format_tmax_measurement(measurement):
	date = format_date(measurement["date"])
	temp = measurement["value"]
	return f"On {date} it was {temp} degrees Fahrenheit"
```

```python
measurement = {
        "date": "2025-07-07T00:00:00",
        "datatype": "TMAX",
        "station": "GHCND:USW00014739",
        "attributes": ",,W,2400",
        "value": 90.0
    }
```

```python
format_tmax_measurement(measurement)
```
