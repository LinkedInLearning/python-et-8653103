```python
p = Path("data_8_2") / "sonnet_tree"
```

```python
for item in p.iterdir():
    print(item.is_dir())
```

```python
sonnets = []

def find_sonnet_files(p):
    for item in p.iterdir():
        if item.is_file():
            sonnets.append(item)
        else:
            find_sonnet_files(item)

find_sonnet_files(p)
```

```python
for sonnet in sonnets[0:10]:
    print(sonnet)
```
